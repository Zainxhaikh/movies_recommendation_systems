# Data Preprocessing

```python
movies = movies[['movie_id', 'title', 'overview', 'genres', 'keywords', 'cast', 'crew']]
movies.head()
```

```python
# Check for missing values
movies.isnull().sum().sort_values(ascending=False)
```

```python
# Impute missing values
movies["overview"] = movies["overview"].fillna(movies["overview"].mode()[0], inplace=True)
movies.isnull().sum().sort_values(ascending=False)
```

```python
# Check for duplicate values
movies.duplicated().sum()
```

```python
# import ast module to convert string representation of list/dict to actual list/dict
import ast
ast.literal_eval("[1, 2, 3]")  # Example usage
```

```python
# funtion to obtain the list of names from the string representation of list of dictionaries
def convert(obj):
    L = []
    for i in ast.literal_eval(obj):
        L.append(i["name"])
    return L
```

```python
# apply the function to the genres column
movies["genres"] = movies["genres"].apply(convert)
```

```python
movies.head()
```

```python
# apply the function to the keywords column for extract the names
movies["keywords"] = movies["keywords"].apply(convert)   
movies.head()
```

```python
# function to get the top 3 cast members from the string representation of list of dictionaries
def convert3(obj):
    L =[]
    counter = 0
    for i in ast.literal_eval(obj):
        if counter < 3:
            L.append(i["name"])
            counter += 1
        else:
            break
    return L
```

```python
# apply the function to the cast column
movies["cast"] = movies["cast"].apply(convert3)
movies.head()
```

```python
# check the column
movies["crew"][0]
```

```python
# function to get the director from the string representation of list of dictionaries
def fetch_director(obj):
    L = []
    for i in ast.literal_eval(obj):
        if i["job"] =="Director":
            L.append(i["name"])
            break
    return L
```

```python
# apply the function to the crew column
movies["crew"] = movies["crew"].apply(fetch_director)
movies.head()
```

```python
# check the overview column
movies["overview"][0]
```

```python
# convert the overview column to a list of words
movies["overview"] = movies["overview"].apply(lambda x: x.split())
movies.head()
```

```python
# remove spaces from the names in the list of genres, keywords, cast and crew
movies["genres"] = movies["genres"].apply(lambda x: [i.replace(" ", "") for i in x])
movies["keywords"] = movies["keywords"].apply(lambda x: [i.replace(" ", "") for i in x])
movies["cast"] = movies["cast"].apply(lambda x: [i.replace(" ", "") for i in x])
movies["crew"] = movies["crew"].apply(lambda x: [i.replace(" ", "") for i in x])
movies.head()
```

```python
# add the new tag column which is a combination of genres, keywords, cast, crew and overview
movies["tags"] = movies["genres"] + movies["keywords"] + movies["cast"] + movies["crew"] + movies["overview"]
movies.head()
```

```python
# create a new dataframe with only the required columns
new_df = movies[["movie_id", "title", "tags"]]
new_df.head()
```

```python
# convert the list of tags to a string
new_df["tags"] = new_df["tags"].apply(lambda x: " ".join(x))
new_df.head()
```

```python
# convert the text of the tags column to lower case
new_df["tags"] = new_df["tags"].apply(lambda x: x.lower())
new_df.head()
```

```python
# import the nltk library for stopwords and stemming
import nltk
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
# create a function to stem the words in the tags column
def stem(text):
    y =[]
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
```

```python
new_df["tags"] = new_df["tags"].apply(stem)
new_df.head()
```

```python
# import the CountVectorizer class from sklearn to convert text data into numerical data
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000, stop_words="english")
```

```python
vectors = cv.fit_transform(new_df["tags"]).toarray()
vectors
```
